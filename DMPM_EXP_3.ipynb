```python
print("\nHistogram")
print("-----------------\n")
plt.figure(figsize=(10, 6))
plt.hist(df['cgpa'], bins=20, edgecolor='black', color='skyblue')
plt.title('Distribution of CGPA', fontsize=14)
plt.xlabel('CGPA', fontsize=12)
plt.ylabel('Number of Students', fontsize=12)
plt.grid(axis='y', alpha=0.75)
plt.show()
```

```python
numeric_df = df.select_dtypes(include=[np.number])

print("\nCorrelation Heatmap (Corrected)")
print("-----------------\n")
plt.figure(figsize=(12, 8))
sns.heatmap(numeric_df.corr(), annot=True, cmap='coolwarm', linewidths=0.5, fmt=".2f")
plt.title('Correlation Heatmap of Numerical Features', fontsize=14)
plt.show()
```

```python
print("\nBar-Plot")
print("-----------------\n")
plt.figure(figsize=(10, 6))
sns.barplot(x='branch', y='cgpa', data=df, palette='viridis')
plt.title('Average CGPA by Branch', fontsize=14)
plt.xlabel('Branch', fontsize=12)
plt.ylabel('Average CGPA', fontsize=12)
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()
```

```python
print("\nPie-Chart")
print("-----------------\n")
plt.figure(figsize=(8, 8))
df['branch'].value_counts().plot(kind='pie', autopct='%1.1f%%', startangle=90, colors=['skyblue', 'lightcoral', 'lightgreen', 'gold', 'lightsalmon'])
plt.title('Distribution of Students by Branch', fontsize=14)
plt.ylabel('')
plt.show()
```

```python
print("\nBox Plot by Branch")
print("-----------------")
plt.figure(figsize=(12, 7))
sns.boxplot(x='branch', y='cgpa', data=df, palette='viridis')
plt.title('CGPA Distribution by Branch', fontsize=14)
plt.xlabel('Branch', fontsize=12)
plt.ylabel('CGPA', fontsize=12)
plt.xticks(rotation=45, ha='right')
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.tight_layout()
plt.show()
```
