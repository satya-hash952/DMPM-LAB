# Without using function Mean, Median, Mode, Variance, Standard Deviation

```python
data=[10,20,30,20,50,20]
f={}
for i in data:
  if i in f:
    f[i]+=1
  else:
    f[i]=1

max_count=max(f.values())

for key,value in f.items():
  if value==max_count:
    mode=key

print("Mode:",mode)
```

```python
data=[10,20,30]

mean=sum(data)/len(data)
t=0
for i in data:
  t+=(i-mean)**2
variance=t/len(data)
print("Variance:",variance)
```

```python
import math
data=[10,20,30]
mean=sum(data)/len(data)
t=0
for i in data:
  t+=(i-mean)**2
variance=t/len(data)
std=math.sqrt(variance)
print("Standard Deviation:",std)
```

# With using function Mean, Median, Mode, Variance, Standard Deviation

```python
def mean(data):
  t= sum(data)

  return t/len(data)

print("Mean:",mean(data))
```

```python
def median(data):
  data.sort()
  n=len(data)
  if(n%2)==0:
    return data[n//2]
  else:
    return(data[n//2-1]+data[n//2])/2

print("Median:",median(data))
```

```python
def mode(data):
  f={}
  for i in data:
    if i in f:
      f[i]+=1
    else:
      f[i]=1

  max_count=max(f.values())

  for key,value in f.items():
    if value==max_count:
      return key

print("Mode:",mode(data))
```

```python
def variance(data):
  mean=sum(data)/len(data)
  t=0
  for i in data:
    t+=(i-mean)**2
  return t/len(data)

print("Variance:",variance(data))
```

```python
import math
def standard_deviation(data):
  mean=sum(data)/len(data)
  t=0
  for i in data:
    t+=(i-mean)**2
  variance=t/len(data)
  std=math.sqrt(variance)
  return std

print("Standard Deviation:",standard_deviation(data))
```
