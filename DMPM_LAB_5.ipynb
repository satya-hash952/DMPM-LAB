# **Association rule mining**

# APRIORI ALGORITHM


1.   Frequent partten mining unsing apriori algorithm



```python
import pandas as pd
from mlxtend.preprocessing import TransactionEncoder
from mlxtend.frequent_patterns import apriori, association_rules
import warnings
warnings.filterwarnings("ignore")
```

```python
dataset = [
    ['Milk', 'Bread', 'Butter'],
    ['Milk', 'Butter', 'Ketchup', 'Eggs'],
    ['Milk', 'Bread', 'Butter', 'Ketchup', 'Eggs'],
    ['Bread', 'Butter', 'Ketchup'],
    ['Milk', 'Bread', 'Ketchup']
]

print("Dataset:")
for transaction in dataset:
    print(transaction)
```

```python
te = TransactionEncoder()
te_ary = te.fit(dataset).transform(dataset)
df = pd.DataFrame(te_ary, columns=te.columns_)

print("--- Binary Encoded DataFrame ---")
print(df)
print("\n" + "="*50 + "\n")
```

```python
frequent_itemsets = apriori(df, min_support=0.6, use_colnames=True)

print("--- Frequent Itemsets Found (Apriori) ---")
print(frequent_itemsets)
print("\n" + "="*50 + "\n")
```

```python
rules = association_rules(frequent_itemsets, metric="confidence", min_threshold=0.7)

print("--- Strong Association Rules ---")
print(rules[['antecedents', 'consequents', 'support', 'confidence', 'lift']])
```
